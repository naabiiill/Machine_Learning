```python
import pandas as pd

pd.read_csv("heart.csv")

print("First 5 rows:")
print(df.head(5))

print("\nLast 5 rows:")
print(df.tail(5))

print("\nShape of dataset:")
print(df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Patients with heart disease:", (df["target"] == 1).sum())
print("Patients without heart disease:", (df["target"] == 0).sum())

print("\nMale patients:", (df["sex"] == 1).sum())
print("Female patients:", (df["sex"] == 0).sum())
```

```python
print("First 10 rows:")
print(df.iloc[:10])

print("\nAge, Sex and Cholesterol:")
print(df.iloc[:, [0, 1, 4]])

print("\nRows 20 to 30 and columns 0 to 4:")
print(df.iloc[20:31, 0:5])
```

```python
print("Patients older than 50:")
older_than_50 = df[df["age"] > 50]
print(older_than_50)

print("\nPatients with heart disease and cholesterol > 240:")
high_risk = df[(df["target"] == 1) & (df["chol"] > 240)]
print(high_risk)

print("\nNumber of patients satisfying both conditions:")
print(len(high_risk))
```
